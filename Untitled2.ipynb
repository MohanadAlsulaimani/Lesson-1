```python
import pandas as pd

# ==========================================
# Task 1: Load the dataset into a dataframe
# ==========================================
# Replace 'student_performance.csv' with the actual filename provided to you
df = pd.read_csv('student_performance.csv')

# Standardize text columns to lowercase to avoid case-sensitivity issues
df['study_hours'] = df['study_hours'].str.lower()
df['attendance'] = df['attendance'].str.lower()
df['pass'] = df['pass'].str.lower()

# Total number of observations (N = 200)
N = len(df)
print(f"Total number of records loaded: {N}\n")

# ==========================================
# Task 2: Calculate individual probabilities
# ==========================================
# Event A: study_hours == "high"
prob_A = len(df[df['study_hours'] == 'high']) / N

# Event B: attendance == "low"
prob_B = len(df[df['attendance'] == 'low']) / N

# Event C: pass == "yes"
prob_C = len(df[df['pass'] == 'yes']) / N

print("--- Task 2: Individual Probabilities ---")
print(f"P(A) [High Study Hours]: {prob_A:.4f}")
print(f"P(B) [Low Attendance]:   {prob_B:.4f}")
print(f"P(C) [Passed Course]:    {prob_C:.4f}\n")

# ==========================================
# Task 3: Calculate joint probabilities
# ==========================================
# a. P(A ∩ B)
prob_A_and_B = len(df[(df['study_hours'] == 'high') & (df['attendance'] == 'low')]) / N

# b. P(B ∩ C)
prob_B_and_C = len(df[(df['attendance'] == 'low') & (df['pass'] == 'yes')]) / N

print("--- Task 3: Joint Probabilities ---")
print(f"P(A ∩ B): {prob_A_and_B:.4f}")
print(f"P(B ∩ C): {prob_B_and_C:.4f}\n")

# ==========================================
# Task 4: Conditional probability P(C|A)
# ==========================================
# P(C|A) = P(A ∩ C) / P(A)
prob_A_and_C = len(df[(df['study_hours'] == 'high') & (df['pass'] == 'yes')]) / N
prob_C_given_A = prob_A_and_C / prob_A

print("--- Task 4: Conditional Probability ---")
print(f"P(C|A): {prob_C_given_A:.4f}\n")

# ==========================================
# Task 5: Test independence of B and C
# ==========================================
# Two events are independent if P(B ∩ C) == P(B) * P(C)
product_prob_B_C = prob_B * prob_C

print("--- Task 5: Independence Test ---")
print(f"P(B ∩ C) = {prob_B_and_C:.4f}")
print(f"P(B) * P(C) = {product_prob_B_C:.4f}")

# Using a small tolerance for floating-point comparison
if abs(prob_B_and_C - product_prob_B_C) < 1e-6:
    print("Conclusion: Events B and C are Independent.")
else:
    print("Conclusion: Events B and C are Dependent (not independent).")
print("\n")

# ==========================================
# Task 6: Apply Bayes' Theorem to find P(A|C)
# ==========================================
# Bayes' Theorem formula: P(A|C) = (P(C|A) * P(A)) / P(C)
prob_A_given_C = (prob_C_given_A * prob_A) / prob_C

print("--- Task 6: Bayes' Theorem ---")
print(f"P(A|C) calculated via Bayes' Theorem: {prob_A_given_C:.4f}")
```
